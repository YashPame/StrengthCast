## 2. Import libraries

````python
import pandas as pd
````

## 3. Set file paths
The notebook is inside `ML/`, so `../data/` points to the `data` folder one level up.  
If your `data` folder is inside `ML/`, change `../data/` to `data/`.

````python
input_path = "../data/Concrete_Data.xls"
output_path = "../data/concrete_data.csv"
````

## 4. Load the Excel file

````python
df = pd.read_excel(input_path)

print("Shape:", df.shape)
df.head()
````

## 5. Look at the original column names
They are long and some contain extra spaces, which makes them hard to use in code.

````python
for col in df.columns:
    print(repr(col))
````

## 6. Rename the columns
We rename the columns by position, so the extra spaces in the original names don't cause problems.

| Original column | New name |
|---|---|
| Cement (component 1)(kg in a m^3 mixture) | `cement` |
| Blast Furnace Slag (component 2)(kg in a m^3 mixture) | `slag` |
| Fly Ash (component 3)(kg in a m^3 mixture) | `fly_ash` |
| Water (component 4)(kg in a m^3 mixture) | `water` |
| Superplasticizer (component 5)(kg in a m^3 mixture) | `superplasticizer` |
| Coarse Aggregate (component 6)(kg in a m^3 mixture) | `coarse_aggregate` |
| Fine Aggregate (component 7)(kg in a m^3 mixture) | `fine_aggregate` |
| Age (day) | `age` |
| Concrete compressive strength(MPa, megapascals) | `strength` |

Units: all ingredients are in **kg/m³**, `age` is in **days**, and `strength` is in **MPa**.

````python
df.columns = [
    "cement",
    "slag",
    "fly_ash",
    "water",
    "superplasticizer",
    "coarse_aggregate",
    "fine_aggregate",
    "age",
    "strength",
]

df.head()
````

## 7. Quick check of the data

````python
df.info()
````

````python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
````

## 8. Save as CSV
`index=False` stops pandas from writing the row numbers as an extra column.

````python
df.to_csv(output_path, index=False)
print("Saved to:", output_path)
````

## 9. Verify the saved file
Load the CSV back to confirm it was saved correctly.

````python
check = pd.read_csv(output_path)
print("Shape:", check.shape)
check.head()
````
